# Exploratory Data Analysis - Global Food Prices
Analysis of WFP Global Food Prices Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('husl')

# Load data
df = pd.read_csv('../data/wfp_food_prices_global_2026.csv')
df['date'] = pd.to_datetime(df['date'])

print(f'Dataset Shape: {df.shape}')
print(f'\nFirst few rows:')
print(df.head())
```

## Data Overview

```python
# Data types and missing values
print('Data Types:')
print(df.dtypes)
print('\nMissing Values:')
print(df.isnull().sum())
print('\nData Info:')
print(df.info())
```

## Descriptive Statistics

```python
# Summary statistics
print('Descriptive Statistics:')
print(df.describe())

# Value counts
print('\nUnique Values:')
print(f'Countries: {df["countryiso3"].nunique()}')
print(f'Commodities: {df["commodity"].nunique()}')
print(f'Categories: {df["category"].nunique()}')
print(f'Markets: {df["market"].nunique()}')
print(f'Currencies: {df["currency"].nunique()}')
```

## Price Analysis

```python
# Price statistics
print('Price Statistics (USD):')
print(df['usdprice'].describe())

print('\nPrice Statistics (Local Currency):')
print(df['price'].describe())

# Price by commodity
print('\nTop 10 Commodities by Average Price:')
print(df.groupby('commodity')['usdprice'].mean().sort_values(ascending=False).head(10))
```

## Category Analysis

```python
# Category breakdown
print('Food Categories:')
print(df['category'].value_counts())

print('\nTop Commodities:')
print(df['commodity'].value_counts().head(15))
```

## Geographic Distribution

```python
# Country analysis
print('Countries in Dataset:')
print(df['countryiso3'].value_counts().head(15))

# Geographic coverage
print('\nGeographic Coordinates Range:')
print(f'Latitude: {df["latitude"].min():.2f} to {df["latitude"].max():.2f}')
print(f'Longitude: {df["longitude"].min():.2f} to {df["longitude"].max():.2f}')
```

## Temporal Analysis

```python
# Date range
print('Date Range:')
print(f'Start: {df["date"].min()}')
print(f'End: {df["date"].max()}')
print(f'Duration: {(df["date"].max() - df["date"].min()).days} days')

# Records per date
print('\nRecords by Date:')
print(df.groupby('date').size().describe())
```

## Data Quality

```python
# Price flags
print('Price Flags (Data Quality):')
print(df['priceflag'].value_counts())

print('\nPrice Types:')
print(df['pricetype'].value_counts())

print('\nUnits Used:')
print(df['unit'].value_counts())
```

## Correlation Analysis

```python
# Correlation matrix
numeric_cols = df.select_dtypes(include=[np.number]).columns
correlation = df[numeric_cols].corr()

fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(correlation, annot=True, fmt='.2f', cmap='coolwarm', center=0, ax=ax)
plt.title('Correlation Matrix', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print('Correlation Matrix:')
print(correlation)
```

## Key Insights

```python
print('KEY INSIGHTS FROM DATA:')
print('='*50)
print(f'\n1. Total Records: {len(df):,}')
print(f'2. Geographic Coverage: {df["countryiso3"].nunique()} countries')
print(f'3. Market Coverage: {df["market"].nunique()} markets')
print(f'4. Food Items: {df["commodity"].nunique()} commodities')
print(f'5. Categories: {df["category"].nunique()} categories')
print(f'\n6. Price Range (USD): ${df["usdprice"].min():.2f} - ${df["usdprice"].max():.2f}')
print(f'7. Average Price (USD): ${df["usdprice"].mean():.2f}')
print(f'8. Price Volatility (Std Dev): ${df["usdprice"].std():.2f}')
print(f'\n9. Time Period: {df["date"].min().date()} to {df["date"].max().date()}')
print(f'10. Missing Values: {df.isnull().sum().sum()} ({df.isnull().sum().sum()/(len(df)*len(df.columns))*100:.2f}%)')
```
